```python
df['order_date']=pd.to_datetime(df['order_date'])
```

```python
df['order_date'].dtype
```

```python
df['month']=df['order_date'].dt.month
```

```python
df['month'].head(10)
```

```python
monthly_revenue=df.groupby('month')['total_revenue'].sum()
```

```python
monthly_revenue
```

```python
april_to_may=((monthly_revenue[5]-monthly_revenue[4])/monthly_revenue[4])*100
```

```python
april_to_may
```

```python
may_to_june=((monthly_revenue[6]-monthly_revenue[5])/monthly_revenue[5])*100
```

```python
may_to_june
```

```python
round(april_to_may,1)
```

```python
round(may_to_june,1)
```

```python
# Q --> 3. The channel decision 
```

```python
channel_revenue=df.groupby('sales_channel')['total_revenue'].sum()
```

```python
channel_revenue
```

```python
channel_revenue.sum()
```

```python
marketplace_share=(channel_revenue['Marketplace']/channel_revenue.sum())*100
```

```python
round(marketplace_share,1)
```

```python
channel_orders=df['sales_channel'].value_counts()
```

```python
channel_orders
```

```python
aov=channel_revenue/channel_orders  # aov = average order value
```

```python
aov
```

```python
round(aov,2)
```

```python
channel_monthly = df.groupby(["month", "sales_channel"])["total_revenue"].sum()
```

```python
channel_monthly
```

```python
channel_monthly.unstack()
```

```python
channel_growth=channel_monthly.unstack().pct_change()*100
```

```python
round(channel_growth,1)
```
